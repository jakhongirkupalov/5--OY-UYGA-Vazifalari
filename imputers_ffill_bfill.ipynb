# Imputerlar  - bu missing valueslarni  (NaN) ni to'ldirish uchun ishlatiladigan metodlar va vositalar

# Imputerlar datasetdagi bosh qiymatni mantiqan toldiradi

# Eng oddiy usullari - Mean//Median//Mode  -- Simple Imputerlari xisoblanadi

```python
import seaborn as sns
from sklearn.impute import SimpleImputer

df = sns.load_dataset('titanic')


imputer = SimpleImputer(strategy='mean')
df['age'] = imputer.fit_transform(df[["age"]])


imputer_2 = SimpleImputer(strategy="most_frequent")
df["embarked"] = imputer_2.fit_transform(df[["embarked"]]).ravel()
```

```python
df.info()
```

```python
imputer_3 = SimpleImputer(strategy="most_frequent")
df["deck"] = imputer_3.fit_transform(df[["deck"]]).ravel()
```

```python
df.info()
```

- tez baseline model qurish uchun
- data normal taqsimlangan bolsa

# KNN Imputer ishlash tartibi 
- eng yaqin qo'shnilar asosida to'ldiradi
- eng o'xshash qatorlarni topadi 
- ularning qiymati asosida NaN larni xisoblaydi


```python
from sklearn.impute import KNNImputer

imputer_KNN = KNNImputer(n_neighbors=5)

df_imputed = imputer_KNN.fit_transform(df.select_dtypes(include="number"))
```

# MICE imputer quyidagicha ishlaydi
- har bir ustunni boshqa ustunlar orqali predict qiladi 
- bir nechta iteratsiya qiladi

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

imputer_MICE = IterativeImputer()
df_imputed = imputer_MICE.fit_transform(df.select_dtypes(include="number"))
```

# Constant Imputer 

# Constant Imputer - doimiy qiymat bilan to'ldiriladi

# Qachon ishlatiladi 
- "unknown", "missing" kabi label kerak bo'lsa ishlatiladi 

```python
imputer = SimpleImputer(strategy="constant", fill_value=0)
df["age"] = imputer.fit_transform(df[["age"]])
```

# Bfill - Backward Fill - bu keyingi (pastdagi) qiymat bilan NaN ni to'ldirib beradi

```python
import pandas as pd
df = pd.DataFrame({"age": [20, None, 13, None, 78]})
```

```python
df["age"] = df["age"].bfill()
```
