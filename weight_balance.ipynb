# Imbalance datasetlarda quyidagicha balanslash holatlarini ishlatamiz

| Holat | Yechim | Tushuntirish |
| :--- | :--- | :--- |
| **Data kam** | SMOTE | Kam sonli klassga sun'iy ma'lumot qo'shadi |
| **Data katta** | Undersampling | Ko'p sonli klass ma'lumotlarini kamaytiradi |
| **Tez yechim** | `class_weight='balanced'` | Ma'lumotga tegmasdan, modelda vazn ajratadi |
| **Deep learning** | Loss weighting | Focal Loss yoki maxsus vaznli Loss funksiya ishlatadi |

```python
import seaborn as sns 
import pandas as pd

df = sns.load_dataset("titanic")

df.head()
```

```python
df["survived"].value_counts()
```

```python
df = df[["pclass", "sex", "age", "fare", "survived"]]
```

```python
df.head()
```

```python
df.info()
```

```python
df["age"] = df["age"].fillna(df["age"].mean())
```

```python
df.info()
```

```python
df["sex"] = df["sex"].map({"male":0, "female":1})
```

```python
df.info()
```

```python
x = df.drop("survived", axis=1)
y = df["survived"]
```

```python
from sklearn.model_selection  import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

# Class Weight No Balanced

```python
from sklearn.tree import DecisionTreeClassifier

model_no_weight = DecisionTreeClassifier(random_state=42)
model_no_weight.fit(x_train,y_train)

y_pred_no = model_no_weight.predict(x_test) 
```

```python
from sklearn.metrics import classification_report

print("without class weight")
print(classification_report(y_test, y_pred_no))
```

# Class Weight Balanced

```python
model_weight = DecisionTreeClassifier(class_weight="balanced", random_state=42)


model_weight.fit(x_train, y_train)

y_pred_weight = model_weight.predict(x_test)
```

```python
print("class_weight='balanced'")
print(classification_report(y_test, y_pred_weight))
```

# Class Weight Custom == o'zimiz Custom xolatda beramiz 

```python
model_custom = DecisionTreeClassifier(class_weight={0:1, 1:3}, random_state=42)

model_custom.fit(x_train, y_train)

y_pred_custom = model_custom.predict(x_test)
```

```python
print("Custom = class_weight")
print(classification_report(y_test, y_pred_custom))
```
